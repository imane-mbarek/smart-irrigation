# Smart Irrigation — Exploratory Analysis


```python
import sys
sys.path.insert(0, '../src')
import pandas as pd
import matplotlib.pyplot as plt
from data.ingestion import load_data
from data.preprocessing import preprocess

df = load_data()
df.describe()
```

```python
fig, axes = plt.subplots(1, 4, figsize=(14, 4))
features = ['soil_moisture', 'temperature', 'humidity', 'rainfall']
colors = {0: '#4ade80', 1: '#f87171'}

for ax, feat in zip(axes, features):
    for label, grp in df.groupby('irrigate'):
        ax.hist(grp[feat], alpha=0.65, bins=20, color=colors[label],
                label='Irrigate' if label else 'No Action')
    ax.set_title(feat.replace('_', ' ').title())
    ax.legend(fontsize=8)

plt.tight_layout()
plt.show()
```

```python
print('Class distribution:')
print(df['irrigate'].value_counts(normalize=True).rename({0:'No Irrigation', 1:'Irrigate'}))
```
